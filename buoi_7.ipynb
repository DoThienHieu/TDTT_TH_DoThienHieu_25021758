```python
#w7a1
def binary_search(a,target):
    start=0
    end=len(a)-1
    while start<end:
        mid=(start+end)//2
        if a[mid]<target:
            start+=1
        elif a[mid]>target:
            end-=1
        else:
            return mid
    return -1
binary_search(list(map(int,input().split())), int(input()))
```

```python
#w7a2
def count_occurrences(arr, x):
    k=arr.count(x)
    return k
lst=list(map(int,input().split()))
x=int(input())
count_occurrences(lst,x)
```

```python
#w7a3
def bubble_sort(a):
    for i in range(len(a)):
        for j in range(len(a)-i-1):
            if a[j]>a[j+1]:
                a[j],a[j+1]=a[j+1],a[j]
    return a
bubble_sort(list(map(int,input().split())))
```
