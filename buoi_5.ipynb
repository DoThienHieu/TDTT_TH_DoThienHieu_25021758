```python
#w5a1
def max_ab(a,b):
    return max(a,b)
a,b=map(int,input().split())
print(max_ab(a,b))
```

```python
#w5a3
def check_prime(n):
    for i in range(2,n):
        if n%i==0:
            return False
    return n>1
print(check_prime(int(input())))
```

```python
#w5a4
def perfect_num(n):
    s=0
    for i in range(1,n):
        if n%i==0:
            s+=i
    if s==n:
        return True
    return False
print(perfect_num(int(input())))
```

```python
#w5a5
def find_k(lst,k):
    for i in range(len(lst)):
        if lst[i]==k:
            return i+1
    return -1
a=list(map(int,input().split()))
k=int(input())
print(find_k(a,k))
```

```python
#w5a6
def gt(n):
    gt=1
    for i in range(1,n+1):
        gt*=i
    return gt
print(gt(int(input())))
```

```python
#w5a7
def tinh_toan(a,b,phep_toan):
    if phep_toan=="+":
        return a+b
    elif phep_toan=="-":
        return a-b
    elif phep_toan=="*":
        return a*b
    elif phep_toan=="/":
        return a/b
    else:
        print("vui long nhap phep toan hop le!")
a,phep_toan,b=input().split()
a=float(a)
b=float(b)
print(f"{tinh_toan(a,b,phep_toan):.2f}")
    
```

```python
#w5a8
def Hamming(a,b):
    c=a^b
    m=bin(c)
    count=0
    for i in m:
        if i=="1":
            count+=1
    return count
a,b=map(int,input().split())
print(Hamming(a,b))
```

```python
#w5a9
def ham(n):
    s=0
    a=str(n)
    for i in a:
        s+=int(i)
    return s
print(ham(int(input())))
```

```python
#w5a10
def ax(a,b):
    d1={}
    d2={}
    for x,y in zip(a,b):
        if x in d1:
            if d1[x]!=y:
                return False
        d1[x]=y
        if y in d2:
            if d2[y]!=x:
                return False
        d2[y]=x
    return True
a,b=input().split()
print(ax(a,b))
```
