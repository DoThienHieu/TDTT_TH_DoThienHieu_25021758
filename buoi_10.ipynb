```python
#w9a1
def merge(a,b):
    i = j = 0
    kq=[]
    while i<len(a) and j<len(b):
        if a[i]<=b[j]:
            kq.append(a[i])
            i+=1
        else:
            kq.append(b[j])
            j+=1
    kq.extend(a[i:])
    kq.extend(b[j:])
    return kq
a=list(map(int,input().split()))
b=list(map(int,input().split()))
merge(a,b)
```

```python
#w9a3
def hoan_doi_min(a):
    so=int(a)
    a=list(a)
    for i in range(len(a)):
        for j in range(i+1,len(a)):
            a[i],a[j]=a[j],a[i]
            if a[0]!=0:
                kq=int("".join(a))
                if int(kq)<so:
                    so=int(kq)
            a[i],a[j]=a[j],a[i]      
    return so
hoan_doi_min(input())
```

```python
#w9a8
def is_path_crossing(moves):
    u=[]
    for i in moves:
        if i=="R":
            u.append((1,0))
        elif i=="L":
            u.append((-1,0))
        elif i=="U":
            u.append((0,1))
        elif i=="D":
            u.append((0,-1))
    x,y=0,0
    diem_da_qua=[]
    diem_da_qua.append((0,0))
    for dx,dy in u:
        x+=dx
        y+=dy
        if (x,y) in diem_da_qua:
            return True
        diem_da_qua.append((x,y))
    return False
print(is_path_crossing(input()))
```

```python
#w9a9
def median(a):
    min1=a[0]
    max1=a[1]
    min2=a[2]
    max2=a[3]
    if a[0]>a[1]:
        min1,max1=a[1],a[0]
    if a[2]>a[3]:
        min2,max2=a[3],a[2]
    # loại bỏ max
    if max1<max2:
        x=max1
    else:
        x=max2
    # loại bỏ min
    if min1<min2:
        y=min2
    else:
        y=min1
    if x>y:
        x,y=y,x # số bé luôn là x, số lớn luôn là y
    if a[4]<x:
        return x
    elif a[4]>y:
        return y
    else:
        return a[4]
median(list(map(int,input().split())))
```

```python
#w9a10
def findRadius(houses, heaters):
    a=sorted(houses)
    b=sorted(heaters)
    i=0
    maxz=0
    for j in range(len(a)):
        h=abs(a[j]-b[i])
        while i+1 < len(b) and abs(a[j]-b[i+1]) <= h:
            i+=1
        maxz=max(maxz,h)
    return maxz
houses=list(map(int,input().split()))
heaters=list(map(int,input().split()))
findRadius(houses,heaters)
```
