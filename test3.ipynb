```python
import math
def check_prime(n):
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return n>1
a,b=map(int, input().split())
for i in (a, b+1):
    if check_prime(i):
        print(i)
```

```python
#Phân tích 1 số thành tích các SNT
def prime(x):
    for i in range(2,x):
        if x%i==0:
            return False
    return x>1
n=int(input())
list=[]
i=2
while n>1:
    if n%i==0 and prime(i):
        list.append(i)
        n//=i
    else:
        i+=1
listnew=[]
for j in list:
    k=str(j)
    listnew.append(k)
kq="x".join(listnew)
print(kq)
    



```

```python
for i in range(n,1,-1):
    if prime(i) and n%i==0:
        list.append(i)
tich=1
for j in list:
    tich*=j
    if tich == n:
        print(list)
    else:
        for k in range(len(list)):
            if tich*list[k]==n:
                list.append(list[k])
        print(f"{n} phan tich thanh tich cac snt la:",*list)
```

```python
#tim so trong ds va tra ve vi tri cua no
def tim_kiem(ds,x):
    for i in range(len(ds)):
        if ds[i]==x:
            return i
    return -1
hieu=[1,3,8,3,6,28,3,4,5]
x=int(input())
tim_kiem(hieu,x)
```
