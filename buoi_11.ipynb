```python
#w10a3
def tong_binh_phuong_cs(n):
    s=0
    for i in str(n):
        s+=int(i)**2
    return s
try:
    n=int(input())
    if n<=0:
        print("No")
    else:
        check=set()
        while n!=1:
            if n in check:
                print("No")
                break
            check.add(n)
            n=tong_binh_phuong_cs(n)
        else:
            print("Yes")
        
except ValueError:
    print("Loi du lieu")
```

```python
#w11a11
def is_valid_regex(a):
    operators=["+","*","?"]
    i=0
    while i<len(a):
        if a[i]=="\\":
            if i+1 >= len(a):
                return False
            i+=2
            continue
        if a[i] in operators:
            # trước nó không được là toán tử hoặc đầu chuỗi
            if i==0 or a[i-1] in operators:
                return False
        i+=1
    return True
n=int(input())
for _ in range(n):
    a=input()
    if is_valid_regex(a):
        print("valid")
    else:
        print("invalid")
#Không bắt đầu bằng * + ?
#Không có * + ? đứng liền nhau
#\ không được đứng cuối chuỗi
```
