```python
#w6a4
s=input().split()
dict={}
for i in s:
    key,value=i.split(":")
    if key not in dict:
        dict[key]=[]
    dict[key].append(value)
print(dict)
```

```python
#w6a5
n=list(map(int,input().split()))
dict={}
duong,am,khong=0,0,0
for i in n:
    if i>0:
        duong+=1
    elif i<0:
        am+=1
    else:
        khong+=1
dict['positives']=duong
dict['negatives']=am
dict['zeros']=khong
print(dict)
```
