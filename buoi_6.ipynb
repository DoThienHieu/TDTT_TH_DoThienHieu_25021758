```python
#w6a4
s=input().split()
dict={}
for i in s:
    key,value=i.split(":")  #tách chuỗi i tại dấu ":"
    if key not in dict:
        dict[key]=[]
    dict[key].append(value)
print(dict)
```

```python
#w6a5
n=list(map(int,input().split()))
dict={}
duong,am,khong=0,0,0
for i in n:
    if i>0:
        duong+=1
    elif i<0:
        am+=1
    else:
        khong+=1
dict['positives']=duong
dict['negatives']=am
dict['zeros']=khong
print(dict)
```

```python
#w6a9
d1=eval(input())
d2=eval(input())
d_new={}
for i in d1:
    if i in d2:
        d_new[i]=d1[i]+d2[i]
    else:
        d_new[i]=d1[i]
for j in d2:
    if j in d1:
        d_new[j]=d1[j]+d2[j]
    else:
        d_new[j]=d2[j]
print(d_new)
```

```python
#w6a10
a=list(map(int,input().split()))
k=int(input())
u=[]
for i in range(len(a)):
    for j in range(i+1,len(a)):
        if a[i]+a[j]==k:
            u.append((a[i],a[j]))
print(sorted(set(u)))
```

```python
#w6a12
def count(a):
    dict={}
    for i in a:
        if i in dict:
            dict[i]+=1
        else:
            dict[i]=1
    return dict
a=list(map(int,input().split()))
x=count(a)
max_count=0
kq=None
for i in x:
    if max_count<x[i]:
        max_count=x[i]
        kq=i
    elif x[i]==max_count and i<kq:
        kq=i
print(kq)
```

```python
#w6a20
def tim_k(lst,k):
    for i in range(len(lst)):
        if lst[i]==k:
            return i+1
    return -1
n=list(map(int,input().split()))
k=int(input())
tim_k(n,k)
```
