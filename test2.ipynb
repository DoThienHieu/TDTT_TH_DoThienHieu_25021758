```python
import math
def check_prime(n):
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return n>1
n=int(input())
for i in range(1,n):
    if check_prime(i):
        print(i, end=" ")
```

```python
def roman_to_int(s):
    roman_map = {
        "I": 1, "V": 5, "X": 10,
        "L": 50, "C": 100,
        "D": 500, "M": 1000
    }

    total = 0
    prev = 0

    # Duyệt từng ký tự từ trái sang phải
    for ch in s:
        value = roman_map[ch]

        # Nếu giá trị hiện tại > giá trị trước đó → trừ đi 2 lần giá trị trước đó
        if prev < value:
            total += value - 2 * prev
        else:
            total += value

        prev = value

    return total

# Ví dụ sử dụng
print(roman_to_int("LVIII"))     # 12     # 900
```
