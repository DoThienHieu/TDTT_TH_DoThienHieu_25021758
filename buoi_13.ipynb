```python
#w11a4
def areAnagrams(path1: str, path2: str) -> bool:
    with open(path1, 'r') as f:
        s1 = f.read().strip()
    with open(path2, 'r') as f:
        s2 = f.read().strip()
    
    return sorted(s1) == sorted(s2)


print(areAnagrams("input.txt","input2.txt"))
```

```python
#w11a8
def fibo_(n:int) -> str:
    if n==0:
        return 'a'
    elif n==1:
        return 'b'
    else:
        return fibo_(n-1) + fibo_(n-2)

def fibo(path:str) -> str:
    with open(path,'r') as f:
        n=int(f.read())
    return fibo_(n)
print(fibo("input.txt"))
```

```python
#w11a10
def getMoney(path: str):
    with open(path, "r", encoding="utf-8") as f:
        return list(map(int, f.read().strip().split()))

def rob(houses):
    if not houses:
        return 0
    if len(houses) == 1:
        return houses[0]
    dp1, dp2 = houses[0], max(houses[0], houses[1])
    for i in range(2, len(houses)):
        dp1, dp2 = dp2, max(dp2, dp1 + houses[i])
    return dp2

path = input().strip()
houses = getMoney(path)
print(rob(houses))
```

```python
#11a12
def maximumProduct(path:str) -> int:
    with open(path,'r') as f:
        n=f.readlines()
    res_fake=[]
    for x in range(1,len(n)):
        res_fake.append(int(n[x].strip()))
    res_riel=[]
    count=3
    while count>0:
        res_riel.append(max(res_fake))
        res_fake.remove(max(res_fake))
        count-=1
    product=1
    for x in res_riel:
        product*=x
    return product

print(maximumProduct("input.txt"))
```

```python
#w11a13
def moveZeroes(fileName:str)->list:
    with open(fileName,'r') as f:
        n=f.readlines()
    res=[]
    for x in n:
        res.append(int(x.strip()))
    for y in range(len(res)):
        if res[y]==0:
            res.pop(y)
            res.append(0)
    return res

print(moveZeroes("input.txt"))
```

```python
#w11a14
def averageTime(path:str) -> int:
    with open(path,'r') as f:
        n=f.readlines()
    time=[]
    for x in range(len(n)):
        y=n[x].split(",")
        time.append(float(y[1].strip()))
        if x==len(n)-1:
            count=y[0]
    sum_time=0
    for x in time:
        sum_time+=x
    average=sum_time/int(count)
    return round(average,2)
print(averageTime("input.txt"))
```

```python
#w11a15
def findMovies(path: str):
    res = []
    with open(path, "r", encoding="utf-8") as f:
        lines = f.read().strip().splitlines()[1:]
        for line in lines:
            parts = [p.strip() for p in line.split(",")]
            movie = parts[1]
            desc = parts[2].lower()
            rating = float(parts[3])
            if rating > 8.0 and desc != "boring":
                res.append(movie)
    if not res:
        return "All are boring!"
    return res

path = input().strip()
result = findMovies(path)
if isinstance(result, list):
    for m in result:
        print(m)
else:
    print(result)
```

```python
#w11a16
import math
path = input().strip()
with open(path, "r", encoding="utf-8") as f:
    lines = [line.strip() for line in f if line.strip() != ""]
i = 0
while i < len(lines):
    typ = lines[i].split(":")[1].strip()
    info = lines[i + 1]
    if typ == "SQUARE":
        side = float(info.split(":")[1])
        p = 4 * side
    elif typ == "RECTANGLE":
        parts = info.split()
        width = float(parts[1])
        height = float(parts[3])
        p = 2 * (width + height)
    elif typ == "CIRCLE":
        r = float(info.split(":")[1])
        p = 2 * math.pi * r
    print(f"{p:.2f}")
    i += 2
```

```python
#w11a17
def productExceptSelf(path: str) -> list[int]:
    with open(path,'r') as f:
        nums = list(map(int, f.read().strip().split()))

    res_real = []

    for i in range(len(nums)):
        prod = 1
        for j in range(len(nums)):
            if i != j:
                prod *= nums[j]
        res_real.append(prod)
    return res_real
print(productExceptSelf("input.txt"))
```

```python
#w11a18
import math

def estimatedTime(path: str) -> float:
    with open(path, 'r') as f:
        lines = f.readlines()

    src = lines[0].split()
    x1 = float(src[2])
    y1 = float(src[4])

    des = lines[1].split()
    x2 = float(des[2])
    y2 = float(des[4])

    vel = lines[2].split()
    v = float(vel[1])

    distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

    time = distance / v

    return round(time, 2)

print(estimatedTime("input.txt"))
```

```python
#w11a20
def imageSmoother(path: str):
    with open(path, "r", encoding="utf-8") as f:
        lines = f.read().strip().splitlines()
        M = int(lines[0])
        a = [list(map(int, lines[i+1].split())) for i in range(M)]

    res = [[0]*M for _ in range(M)]
    for i in range(M):
        for j in range(M):
            s = 0
            cnt = 0
            for dx in [-1, 0, 1]:
                for dy in [-1, 0, 1]:
                    ni, nj = i + dx, j + dy
                    if 0 <= ni < M and 0 <= nj < M:
                        s += a[ni][nj]
                        cnt += 1
            res[i][j] = s // cnt

    for row in res:
        print(*row)

path = input().strip()
imageSmoother(path)
     
```
