```python
#w4a10
def do_dai(n):
    dem=1
    while n != 1:
        if n % 2 == 0:
            n = n/2
        else:
            n = n*3 + 1 
        dem += 1
    return dem
n = int(input("nhap so nguyen n:"))
maxlen = 0 
so = 0
for i in range(1, n+1):
    l = do_dai(i)
    if l > maxlen:
        maxlen=l
        so = i
print(so, maxlen)    
```
