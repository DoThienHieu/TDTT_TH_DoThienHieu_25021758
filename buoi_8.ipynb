```python
#w8a1
class Cylinder:
    def __init__(self,r,h):
        self.h=h
        self.r=r
        self.pi=3.14
    def surface_area(self):
        return 2*self.pi*self.r**2 + 2*self.pi*self.r*self.h
    def volume(self):
        return self.pi*self.r**2*self.h
r,h=map(int,input().split())
cylinder = Cylinder(r,h)
s=cylinder.surface_area()
v=cylinder.volume()
print(f"{s:.2f} {v:.2f}")
```

```python
#w8a2
class Date:
    def __init__(self,d,m,y):
        self.d=d
        self.m=m
        self.y=y
    def leap_year(self):
        if (self.y%4==0 and self.y%100!=0) or self.y%400==0:
            return True
        return False
    def days_in_month(self):
        if self.m==2:
            if self.leap_year():
                return 29
            else:
                return 28
        elif self.m in [1,3,5,7,8,10,12]:
            return 31
        else:
            return 30
    def invalid(self):
        if self.d<1 or self.d>self.days_in_month() :
            return True
        if self.m<1 or self.m>12 or self.y<1:
            return True
        return False
    def next_day(self):
        if self.d==self.days_in_month() and self.m==12:
            self.d=1
            self.m=1
            self.y+=1
        elif self.d==self.days_in_month():
            self.d=1
            self.m+=1
        else:
            self.d+=1
    def print(self):
        return f"{self.d}/{self.m}/{self.y}"
d,m,y=map(int,input().split("/"))
date=Date(d,m,y)
if date.invalid():
    print("INVALID")
else:
    date.next_day()
    print(date.print())
    
            

        
```

```python
#w8a5
class ShoppingCart:
    def __init__(self):
        self.product=[]
    def add(self,name,price):
        self.product.append({"name":name,"price":price})
    def remove(self,name):
        for i in self.product:
            if i["name"]==name:
                self.product.remove(i)
                return True
        return False
    def empty(self):
        if len(self.product)==0:
            return True
        return False
    def total(self):
        total=0
        for i in self.product:
            total+=i["price"]
        return f"tong so tien cac san pham co trong gio hang la: {total}"
    def display(self):
        for i in self.product:
            print(i["name"],i["price"])
    def clear(self):
        self.product.clear()
    def exit(self):
        s=input("ban co muon thoat chuong trinh khong?")
        if s.lower()=="yes":
            return True
        return False
cart = ShoppingCart()
while True:
    print("\n ==== MENU ====")
    print("2. Them vao gio hang nhung san pham moi")
    print("3. Xoa khoi gio hang mot san pham")
    print("4. Kiem tra xem cua hang co rong kh") 
    print("5. Tinh tong toan bo gio hang")
    print("6. Hien thi toan bo san pham trong gio hang") 
    print("7. Xoa toan bo gio hang")
    choice=int(input("chon 1 so tu 2 den 7")) 
    if choice==2:
        a=input()
        b=int(input())
        cart.add(a,b)
        print(f"da them san pham: {a} {b}")
    elif choice==3:
        a=input()
        if cart.remove(a):
            print(f"da xoa san pham: {a}")
        else:
            print("khong tim thay san phan can xoa")
    elif choice==4:
        if cart.empty():
            print("gio hang rong")
        else:
            print("gio hang khong rong")
    elif choice==5:
        print(cart.total())
    elif choice==6:
        cart.display()
    elif choice==7:
        cart.clear()
        print("da xoa toan bo san pham")
    else:
        print("Vui long nhap so tu 2 den 7!")
        continue
    if cart.exit():
        break
    
        
```

```python
#w8a7
import math
class Fraction:
    def __init__(self,num,den):
        self.num=num
        self.den=den
    def simplify(self):
        uc=math.gcd(self.num,self.den)
        self.num//=uc
        self.den//=uc
    def add(self,other):
        num = self.num*other.den + other.num*self.den
        den = self.den*other.den
        return Fraction(num,den)
    def sub(self,other):
        num = self.num*other.den - other.num*self.den
        den = self.den*other.den
        return Fraction(num,den)
    def mul(self,other):
        num = self.num*other.num
        den = self.den*other.den
        return Fraction(num,den)
    def div(self,other):
        num = self.num*other.den
        den = self.den*other.num
        return Fraction(num,den)
    def ans(self):
        self.simplify()
        print(f"{self.num}/{self.den}")
a, b, op, c, d=input().split()
a,b,c,d=map(int,(a,b,c,d))
f1=Fraction(a,b)
f2=Fraction(c,d)

if op=="+":
    ans=f1.add(f2)
elif op=="-":
    ans=f1.sub(f2)
elif op=="*":
    ans=f1.mul(f2)
elif op=="/":
    ans=f1.div(f2)
else:
    print("phep tinh khong hop le")
ans.ans()
```

```python
#w8a8
class BankAccount:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance
    def deposit(self,x):
        self.balance+=x
    def withdraw(self,x):
        if self.balance>=x:
            self.balance-=x
        else:
            print("Insufficient funds")
    def show(self):
        print(f"{self.balance:.2f}")
owner,balance=input().split()
a=float(balance)
bankaccount=BankAccount(owner,a)
n=int(input())
for i in range(n):
    s,k=input().split()
    k=int(k)
    if s=="DEPOSIT":
        bankaccount.deposit(k)
    elif s=="WITHDRAW":
        bankaccount.withdraw(k)
bankaccount.show()
```

```python
#w8a9
class Student:
    def __init__(self,name):
        self.name=name
        self.subjects=[]
        self.scores=[]
    def add(self,subject,score):
        self.subjects.append(subject)
        self.scores.append(score)
    def avg(self):
        if len(self.scores)==0:
            return 0
        else:
            return sum(self.scores)/len(self.subjects)
    def rank(self):
        avg=self.avg()
        if avg >= 8:
            return "Excellent"
        elif avg >= 6.5:
            return "Good"
        elif avg >= 5:
            return "Average"
        else:
            return "Poor"
    def show(self):
        print(f"{self.name} {self.avg():.2f} {self.rank()}")
s=input()
n=int(input())
student=Student(s)
for i in range(n):
    subject, score=input().split()
    student.add(subject,float(score))
student.show()
```

```python
#w8a10
class Book:
    def __init__(self,title, author, year):
        self.title=title
        self.author=author
        self.year=year
class Library:
    def __init__(self):
        self.books=[]
    def add(self,title,author,year):
        self.books.append(Book(title,author,year))
    def count_by_author(self,name):
        count=0
        for i in self.books:
            if i.author==name:
                count+=1
        return count
    def count_by_year(self,year):
        count=0
        for i in self.books:
            if i.year==year:
                count+=1
        return count
lib=Library()
n=int(input())
for _ in range(n):
    s=input().split(" ", 1) # 1 la so lan tach
    a=s[1].split(";")
    if s[0]=="ADD":
        lib.add(a[0],a[1],int(a[2]))
    elif s[0]=="COUNT":
        print(lib.count_by_author(s[1]))
    elif s[0]=="COUNTYEAR":
        print(lib.count_by_year(int(s[1])))
    
```

```python
#w8a11
import math
class Point3D:
    def __init__(self,x,y,z):
        self.x=x
        self.y=y
        self.z=z
    def vector(self,other):
        x=other.x-self.x
        y=other.y-self.y
        z=other.z-self.z
        return Vector3D(x,y,z)
class Vector3D:
    def __init__(self,x,y,z):
        self.x=x
        self.y=y
        self.z=z
    def dot(self,other):
        return self.x*other.x + self.y*other.y + self.z*other.z
    def cross(self,other):
        x=self.y*other.z - other.y*self.z
        y=self.z*other.x - other.z*self.x
        z=self.x*other.y - other.x*self.y
        return Vector3D(x,y,z)
    def norm(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)
A = Point3D(*map(float,input().split()))
B = Point3D(*map(float,input().split()))
C = Point3D(*map(float,input().split()))
D = Point3D(*map(float,input().split()))
AB = A.vector(B)
AC = A.vector(C)
BC = B.vector(C)
BD = B.vector(D)

n1 = AB.cross(AC)
n2 = BC.cross(BD)

cos_phi = n1.dot(n2) / (n1.norm() * n2.norm())
phi = math.acos(cos_phi)*180/math.pi
print(f"{phi:.2f}")

    
```

```python
#w8a12
import math
class Complex:
    def __init__(self,real,imaginary):
        self.real=real
        self.imaginary=imaginary
    def mul(self,other):
        real = self.real*other.real - self.imaginary*other.imaginary
        imaginary = self.imaginary*other.real + self.real*other.imaginary
        return Complex(real,imaginary)
    def mod(self):
        return math.sqrt(self.real**2 + self.imaginary**2)
z1 = Complex(*map(float,input().split()))
z2 = Complex(*map(float,input().split()))
if z1.mod() > 100 or z2.mod() > 100:
    print("so phuc kh hop le")
else:
    a = z1.mul(z2)
    print(a.real,a.imaginary)
```

```python
#w8a13
import math
class Triangle:
    def __init__(self,a,b,c):
        self.a=a
        self.b=b 
        self.c=c
    def area(self):
        p = (self.a + self.b + self.c)/2
        return math.sqrt(p*(p-self.a)*(p-self.b)*(p-self.c))
    def check(self):
        if self.a + self.b <= self.c or self.a + self.c <= self.b  or self.b + self.c <= self.a:
            return True
        return False
triangle = Triangle(*map(int,input().split()))
if triangle.check():
    print("invalid")
else:
    print(round(triangle.area(),2))
```
